```python
def clean_countries(entry):
    parts = [name.strip() for name in entry.split(",")]
    parts = [name for name in parts if name != ""]
    return ", ".join(parts)
```

```python
clean_countries(", France, Algeria")
```

```python
clean_countries("France, Belgium, Luxembourg, Cambodia,")
```

```python
df_clean["country_clean"] = df_clean["country"].apply(clean_countries)
```

```python
generic_labels = {"Movies", "TV Shows"}

def clean_genres(entry):
    parts = [g.strip() for g in entry.split(",")]
    parts = [g for g in parts if g != "" and g not in generic_labels]
    return ", ".join(parts) if parts else np.nan
```

```python
df_clean["listed_in_clean"] = df_clean["listed_in"].apply(clean_genres)
```

```python
conditions = [
    (df_clean["duration_unit"] == "min") & (df_clean["duration_value"] < 40),
    (df_clean["duration_unit"] == "min") & (df_clean["duration_value"] <= 120),
    (df_clean["duration_unit"] == "min"),
    (df_clean["duration_value"] == 1),
    (df_clean["duration_value"] <= 3),
]
labels = ["Short film (<40 min)", "Standard (40-120 min)", "Long (>120 min)", "1 season", "2-3 seasons"]
df_clean["duration_category"] = np.select(conditions, labels, default="4+ seasons")
```
